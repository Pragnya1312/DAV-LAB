```python
# Importing required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the dataset
df = pd.read_csv(r"C:\Users\vbpat\Downloads\archive (1)\penguins_size.csv")

# Displaying first 5 rows
print(df.head())

# Displaying last 5 rows
print(df.tail())

# Displaying dataset shape
print(df.shape)

# Displaying column names
print(df.columns)

# Displaying dataset information
print(df.info())

# Checking missing values
print(df.isnull().sum())

# Removing missing values
df = df.dropna()

# Checking duplicate values
print("Duplicate values:", df.duplicated().sum())

# Removing duplicate values
df = df.drop_duplicates()

# Statistical description
print(df.describe())

# Setting plot style
sns.set_style("whitegrid")

# Histogram
plt.figure(figsize=(8,5))
sns.histplot(df['body_mass_g'], kde=True)
plt.title("Distribution of Penguin Body Mass")
plt.show()

# Box plot
plt.figure(figsize=(8,5))
sns.boxplot(x='species', y='body_mass_g', data=df)
plt.title("Penguin Body Mass by Species")
plt.show()

# Scatter plot
plt.figure(figsize=(8,5))
sns.scatterplot(x='bill_length_mm', y='bill_depth_mm',
                hue='species', data=df)
plt.title("Bill Length vs Bill Depth")
plt.show()

# Pair plot
sns.pairplot(df, hue='species')
plt.show()

# Correlation matrix
numeric_df = df.select_dtypes(include='number')
corr = numeric_df.corr()

# Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
